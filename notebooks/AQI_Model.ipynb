# AQI Dataset - Exploratory Data Analysis
Run this from the `notebooks/` folder (or adjust the path below).

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style('whitegrid')
df = pd.read_csv('../dataset/air_quality.csv')
df.head()
```

```python
df.info()
df.describe()
```

## Missing values & duplicates

```python
print(df.isna().sum())
print('Duplicate rows:', df.duplicated().sum())
```

## Distributions

```python
df.hist(figsize=(14, 10), bins=30)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(14, 8))
df.boxplot(rot=45)
plt.title('Boxplots of All Features')
plt.show()
```

## Correlation heatmap

```python
plt.figure(figsize=(10, 8))
sns.heatmap(df.corr(numeric_only=True), annot=True, fmt='.2f', cmap='coolwarm')
plt.title('Feature Correlation Matrix')
plt.show()
```

## AQI distribution & pair plot

```python
plt.figure(figsize=(8, 5))
sns.histplot(df['AQI'], bins=40, kde=True)
plt.title('AQI Distribution')
plt.show()
```

```python
sns.pairplot(df[['PM2.5', 'PM10', 'NO2', 'AQI']].sample(min(1000, len(df))))
plt.show()
```
